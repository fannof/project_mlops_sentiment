```python
import base64
import json
import tensorflow as tf

def prepare_json(text):
    # Mengonversi input text menjadi tf.train.Example yang diserialisasi
    feature_spec = {
        "ulasan": tf.train.Feature(bytes_list=tf.train.BytesList(value=[bytes(text, "utf-8")]))
    }
    
    example = tf.train.Example(features=tf.train.Features(feature=feature_spec)).SerializeToString()

    # Menyandikan contoh menjadi base64 untuk dikirimkan melalui JSON
    result = {
        "instances": [{"examples": {"b64": base64.b64encode(example).decode()}}]
    }
    
    return json.dumps(result)

# Teks contoh yang akan dikirim
text_test = "Aplikasi yang sangat bagus"
json_data = prepare_json(text_test)

# Mengirimkan permintaan prediksi ke model
endpoint = "http://localhost:8080/v1/models/sentiment-detection-model:predict"
response = requests.post(endpoint, data=json_data, headers={"Content-Type": "application/json"})

# Cetak respons untuk debugging
print(response.text)

# Mendapatkan prediksi
prediction = response.json().get("predictions")
if prediction:
    prediction_value = prediction[0][0]
    result = "Positif" if prediction_value > 0.5 else "Negatif"
else:
    result = "Error: No predictions found in response."

print(result)
```

Dalam melakukan testing, model berhasil memprediksi kalimat "Aplikasi yang sangat bagus" termasuk dalam sentimen positif, dan memang benar kata tersebut adalah sentimen positif.
